```python
import pandas as pd
import numpy as np
data = {
    "Student": ["A","B","C","D","E"],
    "Gender": ["Male","Female","Male","Female","Male"],
    "StudyHours": [5,6,np.nan,4,7],
    "Attendence": [80,85,75,np.nan,90],
    "Marks": [75,82,68,55,np.nan],
    "Result": ["Pass","Pass","Pass","Fail","Pass"]
}
df = pd.DataFrame(data)
print(df)
```

```python
#missing values
print(df.isnull().sum())
```

```python
#mean impution
df["StudyHours"]=df["StudyHours"].fillna(df["StudyHours"].mean())
print(df)
```

```python
#median impution
df["Marks"]=df["Marks"].fillna(df["Marks"].median())
print(df)
```

```python
#forward fill
df["Attendence"] = df["Attendence"].ffill()
print(df)
```

```python
df["Gender"]=df["Gender"].map({
    "Male":1,
    "Female":0
})
print(df)
```

```python
df["Result"]=df["Result"].map({
    "Pass":1,
    "Fail":0
})
print(df)
```

```python
#to check the missing values
print(df.isnull().sum())
```
